```python
import numpy as np

# --- Part 1: 1D Array Slicing ---
arr = np.array([1, 2, 3, 4, 5])  # 1D array
print(arr)

print(arr[0])  # Output: 1
print(arr[0:2])  # Output: [1 2]
print(arr[0:5:2])  # Output: [1 3 5] (Starting index : ending index : step)
print(arr[:5])  # Output: [1 2 3 4 5]

# --- Part 2: 2D Array Slicing ---
arr1 = np.array([[1, 2, 3], [3, 4, 5], [7, 8, 9]])  # 2D Array
print(arr1)

print(arr1[0, 1])  # Row index 0, Col index 1 -> Output: 2
print(arr1[:, 2])  # All rows, Col index 2 -> Output: [3 5 9]
print(arr1[0:2, 0:2])  # First 2 rows and first 2 columns -> Output: [[1 2], [3 4]]
print(
    arr1[0:3:2, 0:3:2]
)  # Step 2 slicing -> Output: [[1 3], [7 9]] (Row 0, 2 & Col 0, 2)
print(arr1[0:2, 1:3])  # First 2 rows and last 2 columns -> Output: [[2 3], [4 5]]

# --- Part 3: Reshape Operations ---
arr_reshape = np.array([1, 2, 3, 4, 5, 6])  # 1D array with 6 elements
print(arr_reshape)

# 2 Rows, 3 Columns
newarr1 = arr_reshape.reshape(2, 3)
print(newarr1)

# 3 Rows, 2 Columns
newarr2 = arr_reshape.reshape(3, 2)
print(newarr2)

# 1 Row, 6 Columns (2D array format)
newarr3 = arr_reshape.reshape(1, 6)
print(newarr3)

# --- Part 4: 3D Reshape Fix ---
# 6 elements cannot be reshaped to (3, 2, 2) which requires 3*2*2 = 12 elements.
# Changing to (3, 2, 1) or changing the input array size to 12.
ar1 = arr_reshape.reshape(3, 2, 1)  # 3 blocks, 2 rows, 1 column
print(ar1)
```

```python
import numpy as np

marks = np.array([55, 60, 72, 80, 85, 90, 95])

print("Mean:", np.mean(marks))
print("Median:", np.median(marks))
print("Std:", np.std(marks))
print("Q1:", np.percentile(marks, 25))
print("Q3:", np.percentile(marks, 75))
```

```python
#print the outlier using the iqr range

import numpy as np

marks = np.array([55, 60, 72, 80, 85, 90, 95])

# Calculate Q1 and Q3
Q1 = np.percentile(marks, 25)
Q3 = np.percentile(marks, 75)

# Calculate IQR
IQR = Q3 - Q1
print("IQR:", IQR)

# Calculate outlier boundaries
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)

# Find outliers
outliers = marks[(marks < lower_bound) | (marks > upper_bound)]

print("Outliers:", outliers)
```

```python
num_features = len(features)
distance_matrix = np.zeros((num_features, num_features))

for i in range(num_features):
    for j in range(num_features):
        distance = np.sum(np.abs(features[i] - features[j]))
        distance_matrix[i][j] = distance

print(distance_matrix)
```

```python
import numpy as np

num_features = len(features)
distance_matrix = np.zeros((num_features, num_features))

for i in range(num_features):
    for j in range(num_features):
        # Euclidean Distance: np.sqrt(np.sum((a - b)**2)) or np.linalg.norm(a - b)
        distance = np.sqrt(np.sum((features[i] - features[j]) ** 2))
        distance_matrix[i][j] = distance

print(distance_matrix)
```

```python
import numpy as np
import pandas as pd

df = pd.DataFrame({
    'Student': ['A', 'B', 'C', 'D'],
    'Math': [80, 82, 40, 20],
    'CSE': [70, 72, 50, 25],
    'English': [90, 88, 45, 15],
    'Grade': ['A', 'A', 'B', 'C'],
})

num_students = len(df)
nominal_matrix = np.zeros((num_students, num_students))
grades = df['Grade'].values

for i in range(num_students):
    for j in range(num_students):

        nominal_matrix[i][j] = 1 if grades[i] != grades[j] else 0

print('Nominal Distance Matrix (Grade):')
print(
    pd.DataFrame(
        nominal_matrix, index=df['Student'], columns=df['Student']
    )
)
```

```python
# Rank mapping for Ordinal Grade
grade_rank = {'A': 0, 'B': 1, 'C': 2}
ranks = df['Grade'].map(grade_rank).values
max_rank = max(grade_rank.values()) - min(grade_rank.values())  # 2 - 0 = 2

ordinal_matrix = np.zeros((num_students, num_students))

for i in range(num_students):
    for j in range(num_students):

        ordinal_matrix[i][j] = abs(ranks[i] - ranks[j]) / max_rank

print('Ordinal Distance Matrix (Grade):')
print(
    pd.DataFrame(
        ordinal_matrix, index=df['Student'], columns=df['Student']
    )
)
```
