```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'Date': [
        '2026-01-05', '2026-01-10', '2026-01-15',
        '2026-02-03', '2026-02-12', '2026-02-20',
        '2026-03-02', '2026-03-11', '2026-03-18',
        '2026-04-04', '2026-04-14', '2026-04-25'
    ],
    'Product': [
        'Laptop', 'Mobile', 'Tablet',
        'Laptop', 'Mobile', 'Headphones',
        'Tablet', 'Laptop', 'Mobile',
        'Headphones', 'Tablet', 'Laptop'
    ],
    'Region': [
        'Dhaka', 'Chittagong', 'Sylhet',
        'Dhaka', 'Khulna', 'Dhaka',
        'Chittagong', 'Sylhet', 'Dhaka',
        'Khulna', 'Dhaka', 'Chittagong'
    ],
    'Quantity': [
        5, 10, 8, 4, 12, 15, 6, 7, 9, 20, 10, 6
    ],
    'Unit_Price': [
        80000, 25000, 30000, 80000, 25000, 3000,
        30000, 80000, 25000, 3000, 30000, 80000
    ]
}

df = pd.DataFrame(data)
df['Total_Sales'] = df['Quantity'] * df['Unit_Price']
df['Date'] = pd.to_datetime(df['Date'])

# Make sublet
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# 1. Product vs Quantity bar chart
prod_qty = df.groupby('Product')['Quantity'].sum()
ax1.bar(prod_qty.index, prod_qty.values, color='skyblue')
ax1.set_title('Product-wise Quantity Sold')
ax1.set_xlabel('Product')
ax1.set_ylabel('Total Quantity')

# 2. sales trend by days (Line Chart)
daywise = df.groupby('Date')['Total_Sales'].sum().reset_index()
ax2.plot(daywise['Date'], daywise['Total_Sales'], marker='o', color='green', linestyle='-', linewidth=2)
ax2.set_title('Day-wise Sales Trend')
ax2.set_xlabel('Date')
ax2.set_ylabel('Total Sales (BDT)')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
